```python
x = df.drop(['departure_delay_minutes'],axis=1)
y = df['departure_delay_minutes']
x
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,random_state=42,test_size=0.2)
```

```python
from sklearn.linear_model import LinearRegression

lr = LinearRegression()

lr.fit(x_train,y_train)

predict_y1 = lr.predict(x_test)

from sklearn.metrics import r2_score , mean_squared_error

print(r2_score(y_test , predict_y1))
print(mean_squared_error(y_test , predict_y1))
```

```python
from sklearn.tree import DecisionTreeRegressor

dr = DecisionTreeRegressor()

dr.fit(x_train,y_train)

predict_y2 = dr.predict(x_test)

from sklearn.metrics import r2_score , mean_squared_error

print(r2_score(y_test , predict_y2))
print(mean_squared_error(y_test , predict_y2))
```

```python
from sklearn.svm import SVR

svv = SVR()

svv.fit(x_train,y_train)

predict_y3 = svv.predict(x_test)

from sklearn.metrics import r2_score , mean_squared_error

print(r2_score(y_test , predict_y3))
print(mean_squared_error(y_test , predict_y3))
```

```python
from sklearn.neighbors import KNeighborsRegressor

knn = KNeighborsRegressor()
knn.fit(x_train, y_train)
predict_y4 = knn.predict(x_test)

from sklearn.metrics import mean_squared_error, r2_score

print(r2_score(y_test, predict_y4))
print(mean_squared_error(y_test, predict_y4))
```

```python
import joblib

joblib.dump(lr, 'flight_delay_model.pkl')

model = joblib.load('flight_delay_model.pkl')
```
